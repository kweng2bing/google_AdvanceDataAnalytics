## Q-Q Plot (Quantile-Quantile Plot)



## Purpose

* A graphical method to determine whether a dataset follows a specific probability distribution
* Plots sample quantiles against theoretical quantiles


### Quantiles
* Points in a dataset that divide the data into intervals containing equal probabilities/proportions of the total distributions.
* Used to describe the spread of data set

#### Examples of Quantiles
* Median
* Quartiles
* Percentiles
